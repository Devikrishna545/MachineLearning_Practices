```python
import torch
```

In Pytorch partial differentiation

```python
x= torch.tensor(.0,requires_grad=True)
x
```

```python
y=torch.tensor(.0,requires_grad=True)
y
```

```python
def f(my_x,my_y):
  return my_x**2 - my_y**2
```

```python
z=f(x,y) #forward pass
```

```python
z.backward() #autodiff
```

```python
x.grad
```

```python
y.grad
```

```python
#x=3,y=0

x=torch.tensor(3.,requires_grad=True)
y=torch.tensor(0.,requires_grad=True)
```

```python
z=f(x,y)
```

```python
z.backward()
```

```python
x.grad
```

```python
y.grad
```

In Tensorflow

```python
import tensorflow as tf
```

```python
from numpy import dtype
x=tf.Variable(3.0,dtype =tf.float32)
y=tf.Variable(0.0,dtype=tf.float32)
```

```python
with tf.GradientTape(persistent=True) as tape:
  z=f(x,y)
```

```python
dz_dx=tape.gradient(z,x)
dz_dx
```

```python
dz_dy = tape.gradient(z,y)
dz_dy
# Remember to call del tape when you are done with all gradient calculations.
# For example: del tape
```

```python
del tape
```
