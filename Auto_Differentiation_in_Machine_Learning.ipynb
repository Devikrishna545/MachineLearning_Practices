#Auto differentiation using Pytorch

Here we can do autodifferentiation / auto diff using these libraries.
y=x^2 ,when x= 5 , find dy/dx?

dy/dx = 2*x =5*2 =10

Now i have to do it using auto diff method.

```python
#import pytorch
import torch
```

```python
#to store value of x as scalar with tensor.
x=torch.tensor(5.0)
x.requires_grad_() # makes requires_grad =  'True' ,it tracks the gradients wherever the x changes in the forward pass.
```

```python
#now y =x^2
y = x**2
```

```python
y.backward() #to find dy
```

```python
x.grad # to find slop
```

# Auto Differentiation using TensorFlow

```python
#import tensor flow
import tensorflow as tf
```

```python
#to store variable
x=tf.Variable(5.0)
with tf.GradientTape() as t:
  t.watch(x) #track forward
  y= x**2
```

```python
t.gradient(y,x) #use autodiff with tensorflow gradient function
```
